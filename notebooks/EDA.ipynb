# Analisis exploratorio de datos

Este notebook contiene una exploracion inicial del dataset `vehicles_us.csv` usando pandas y Plotly.

```python
from pathlib import Path

import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
```

## Cargar los datos

La ruta se resuelve de forma flexible para que el notebook funcione desde VS Code, ya sea que el directorio de trabajo sea la raiz del proyecto o la carpeta `notebooks`.

```python
data_path = Path("../data/vehicles_us.csv")

if not data_path.exists():
    data_path = Path("data/vehicles_us.csv")

car_data = pd.read_csv(data_path)
car_data.head()
```

## Histograma del odometro

Este grafico ayuda a observar la distribucion del kilometraje registrado en los anuncios.

```python
fig = go.Figure(data=[go.Histogram(x=car_data["odometer"])])
fig.update_layout(
    title_text="Distribucion del odometro",
    xaxis_title="Odometro",
    yaxis_title="Cantidad de anuncios",
)
fig.show()
```

## Relacion entre odometro y precio

El grafico de dispersion permite explorar si los vehiculos con mayor kilometraje tienden a tener precios distintos.

```python
scatter_data = car_data.dropna(subset=["odometer", "price"])

fig = go.Figure(
    data=[
        go.Scatter(
            x=scatter_data["odometer"],
            y=scatter_data["price"],
            mode="markers",
            marker={"opacity": 0.35, "size": 5},
        )
    ]
)
fig.update_layout(
    title_text="Relacion entre odometro y precio",
    xaxis_title="Odometro",
    yaxis_title="Precio",
)
fig.show()
```

## Precio por condicion

Este boxplot compara la distribucion de precios entre condiciones reportadas para los vehiculos.

```python
fig = px.box(
    car_data,
    x="condition",
    y="price",
    title="Distribucion del precio por condicion",
)
fig.update_layout(xaxis_title="Condicion", yaxis_title="Precio")
fig.show()
```
